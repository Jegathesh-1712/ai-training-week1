```python
numbers = list(range(10000))

def using_list_comprehension():
    return [x * 2 for x in numbers]

def using_for_loop():
    result = []
    for x in numbers:
        result.append(x * 2)
    return result
```

```python
%timeit using_list_comprehension()
```

```python
%timeit using_for_loop()
```
