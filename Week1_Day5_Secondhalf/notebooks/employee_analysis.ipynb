```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load employee dataset
df = pd.read_csv("../data/employees.csv")

# Display dataset
print(df)

# Dataset information
print("\nDataset Information:")
print(df.info())

# Basic statistics
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Chart 1: Average Salary by Department

import os

# Create figures folder if it doesn't exist
os.makedirs("../reports/figures", exist_ok=True)

# Calculate average salary by department
avg_salary = df.groupby("department")["salary"].mean()

# Create bar chart
plt.figure(figsize=(9, 5))

avg_salary.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black"
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# Save chart
plt.savefig("../reports/figures/bar_chart.png", dpi=300)

# Display chart
plt.show()

print("Insight: The Finance department has the highest average salary.")
```

```python
# Chart 2: Salary Distribution Histogram

plt.figure(figsize=(9, 5))

plt.hist(
    df["salary"],
    bins=5,
    color="orange",
    edgecolor="black"
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary (INR)")
plt.ylabel("Number of Employees")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# Save histogram
plt.savefig("../reports/figures/histogram.png", dpi=300)

plt.show()

print("Insight: The histogram shows how employee salaries are distributed across different salary ranges.")
```

```python
# Chart 2: Employee Salary Distribution

plt.figure(figsize=(9, 5))

plt.hist(
    df["salary"],
    bins=5,
    color="orange",
    edgecolor="black"
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary (INR)")
plt.ylabel("Number of Employees")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

plt.savefig("../reports/figures/histogram.png", dpi=300)

plt.show()

print("Insight: This histogram shows the distribution of employee salaries across different salary ranges.")
```

```python
# Chart 3: Employee ID vs Salary Scatter Plot

plt.figure(figsize=(9, 5))

plt.scatter(
    df["id"],
    df["salary"],
    color="green",
    marker="o",
    label="Employees",
    s=100
)

plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary (INR)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()

# Save scatter plot
plt.savefig("../reports/figures/scatter_plot.png", dpi=300)

plt.show()

print("Insight: The scatter plot shows salary differences among employees.")
```

```python
# Chart 4: Employee Salary Sorted Line Chart

sorted_salary = df["salary"].sort_values().reset_index(drop=True)

plt.figure(figsize=(9, 5))

plt.plot(
    sorted_salary.index + 1,
    sorted_salary,
    color="red",
    marker="o",
    linewidth=2,
    label="Employee Salary"
)

plt.title("Employee Salary Trend (Sorted)")
plt.xlabel("Employee Rank (Lowest to Highest Salary)")
plt.ylabel("Salary (INR)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()

# Save line chart
plt.savefig("../reports/figures/line_chart.png", dpi=300)

plt.show()

print("Insight: Employee salaries increase from the lowest to the highest value.")
```

```python
# Chart 5: Employee Salary Box Plot

plt.figure(figsize=(9, 5))

sns.boxplot(
    y=df["salary"],
    color="lightgreen"
)

plt.title("Employee Salary Distribution - Box Plot")
plt.xlabel("Employee Salary")
plt.ylabel("Salary (INR)")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# Save box plot
plt.savefig("../reports/figures/box_plot.png", dpi=300)

plt.show()

print("Insight: The box plot shows the median salary, salary spread and possible outliers.")
```
