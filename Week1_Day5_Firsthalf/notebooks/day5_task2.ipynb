```python
import pandas as pd
import numpy as np

# Load cleaned employee dataset
df = pd.read_csv("../data/clean_data.csv")

# Display dataset
display(df.head())

print("Dataset Shape:", df.shape)
```

```python
# Summary statistics for numeric columns
summary_stats = df.describe().T

display(summary_stats)
print("Median Salary:", df["salary"].median())
```

```python
department_analysis = df.groupby("department").agg(
    Average_Salary=("salary", "mean"),
    Maximum_Salary=("salary", "max"),
    Minimum_Salary=("salary", "min"),
    Employee_Count=("id", "count")
)

display(department_analysis)
```

```python
# City-wise employee analysis

city_analysis = df.groupby("city").agg(
    Average_Salary=("salary", "mean"),
    Maximum_Salary=("salary", "max"),
    Employee_Count=("id", "count")
)

display(city_analysis)
```

```python
# Find correlations between numeric columns

correlation = df.corr(numeric_only=True)

print("Correlation Matrix:")
display(correlation)
```

```python
# Find top correlation relationships

corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) <
    corr_pairs.index.get_level_values(1)
]

top_relationships = corr_pairs.abs().sort_values(
    ascending=False
)

print("Top Correlation Relationships:")
display(top_relationships.head(3))
```

```python
# Business Question 1:
# Which department has the highest average salary?

highest_salary = department_analysis["Average_Salary"].idxmax()
highest_value = department_analysis["Average_Salary"].max()

print("Q1: Which department has the highest average salary?")
print(f"Answer: {highest_salary} with salary {highest_value:.2f}")


# Business Question 2:
# Which department has the most employees?

largest_department = department_analysis["Employee_Count"].idxmax()

print("\nQ2: Which department has the most employees?")
print(f"Answer: {largest_department} with {department_analysis.loc[largest_department, 'Employee_Count']} employees")


# Business Question 3:
# Which department has the lowest average salary?

lowest_salary = department_analysis["Average_Salary"].idxmin()
lowest_value = department_analysis["Average_Salary"].min()

print("\nQ3: Which department has the lowest average salary?")
print(f"Answer: {lowest_salary} with salary {lowest_value:.2f}")
```

```python
# Load department budget dataset
budget_df = pd.read_csv("../data/department_budget.csv")

print("Department Budget:")
display(budget_df)

# Merge employee and budget datasets
merged_df = pd.merge(
    df,
    budget_df,
    on="department",
    how="left"
)

print("Merged Employee and Budget Data:")
display(merged_df)
```

```python
# Bonus Challenge: Create Pivot Table

pivot_table = pd.pivot_table(
    df,
    values="salary",
    index="department",
    columns="city",
    aggfunc="mean"
)

print("Department-wise and City-wise Average Salary:")
display(pivot_table)
```

```python
# Save merged employee dataset

merged_df.to_csv(
    "../data/merged_employee_data.csv",
    index=False
)

print("Merged dataset saved successfully!")

display(merged_df.head())
```
